### Conditionals

When we want to execute a command/command block only if a certain condition is satisfied, we use if/elif/else.

if < a boolean expression >:

&emsp;< block >

First the Boolean expression is evaluated and the block of code is executed only if the condition is True.


````python
age_Maria = 5
age_Joao = 40
if age_Maria > 18 and age_Joao > 18:
  print("Both are adults.")
elif age_Maria > 18 or age_Joao > 18:
  print("At least one of them is an adult.")
else:
  print("None of them are adults.")
````

## 6. Loops


### While Loops

while < a boolean expression >:

&emsp;< block >

While is, in its logic, very similar to an if statement. The difference is, after we execute the code inside while, instead of continuing the
code, we go back and check if the Boolean expression still True. If it is still True, we execute the code inside while again, and then check the Boolean expression again. We continue this until the Boolean expression is False.

````python
i = 0
while i < 3:
    print("hello")
    i = i + 1
````

### For Loops

for < variable > in < sequence >:

&emsp;< block >


The value of the variable changes in every iteration.


Sequence can be:
1) A string (sequence of characters)
2) A list (sequence of elements)
3) A range (sequence of numbers)

````python
my_str = "hello"
for letter in my_str:
    print(letter)
````

## 7. Functions

Functions are blocks of code we define once but can use multiple times. They are used
to prevent repeating code and to organize the structure of our program.

More specifically, they are blocks of code that are only executed when you call them (i.e.,
when you call the function)

```python
def function_name (input):
    instructions
    return output
```

````python
# no input, no output
def my_func():
    print("hello world")
````

````python
# calling the function
my_func()
````

````python
# input, no output
def my_func(a):
    print("your input is", a)

my_func(6)
````

````python
# input, output
def my_func(a, b):
    return a + b

my_func(3, 4)
````

A function can be thought of as a set of rules mapping inputs to corresponding outputs like a mathematical function mapping one domain to another. Functions can be *pure* and *impure*:
- **Pure** functions **only** operate on explicit inputs, without side effects. 
- **Impure** functions have side effects, which are not related to the inputs, such as modifications to global variables or accessing (read/write) the filesystem. 

````python
# Pure: returns a new list
def add_to_list_pure(lst, item):
    return lst + [item]

# Impure: mutates the input list
def add_to_list(lst, item):
    lst.append(item)
    return lst
````

````python
xs2 = [1, 2]
zs = add_to_list_pure(xs2, 3) # xs2 not mutated
xs2
````

````python
xs = [1, 2]
ys = add_to_list(xs, 3) # xs mutated
xs
````
