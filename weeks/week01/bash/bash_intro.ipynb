# Bash and python

There are two ways we can interact with bash from within our jupyter notebook, first we can use magics.


Magics allow us to write different types of code, for example bash. Run the cell below to see a list of magics:

```python
%lsmagic
```

When we start a cell with %% the whole cell will be run as a magic. Note you cannot combine bash and python code in a single cell in this way. so for example we start a cell with %%bash to run only bash code in that cell:

```python
%%bash
pwd
```

```python
import os
os.getcwd()
```

```python
%%bash
mkdir bashtest
```

```python
%%bash
ls
```

```python
%%bash
cd bashtest
```

```python
%%bash
pwd
```

We can also just use the ! at the start of bash a single bash line (again you cannot mix bash and python in a single cell):

```python
!dir
```

```python
!wsl ls
```
